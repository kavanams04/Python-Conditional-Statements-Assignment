Q2. Explain the difference between if, if-else and if-elif-else.

if: Executes code when the condition is true.
if-else: Executes one block if the condition is true and another if false.
if-elif-else: Checks multiple conditions and executes the matching block.

```python
# Demonstration of if, if-else and if-elif-else

marks = 75

if marks >= 35:
    print("Pass")

if marks >= 35:
    print("You passed")
else:
    print("You failed")

if marks >= 90:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
else:
    print("Grade C")
```

Q3. Write a program to check whether a number is positive, negative or zero.

Steps:
1. Get a number from the user.
2. Check whether it is greater than zero.
3. Check whether it is less than zero.
4. Otherwise, it is zero.

```python
# Program to check positive, negative or zero

num = int(input("Enter a number: "))

if num > 0:
    print("Positive number")
elif num < 0:
    print("Negative number")
else:
    print("Zero")
```

Q4. Write a program to check whether a number is even or odd.

Steps:
1. Enter a number.

3. If the remainder is zero, it is even.
4. Otherwise, it is odd.

```python
# Program to check even or odd number

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

Q5. Write a program to find the greatest among three numbers.

Steps:
1. Enter three numbers.
2. Compare the first number with the other two.
3. If it is greatest, print it.
4. Otherwise, compare the second and third numbers.

```python
# Program to find greatest among three numbers

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    print("Greatest number:", a)
elif b >= a and b >= c:
    print("Greatest number:", b)
else:
    print("Greatest number:", c)
```

Q6. Write a program to check whether a given year is a leap year.

A leap year has 366 days instead of 365 days.

Steps:
1. Enter a year.
2. Check whether it is divisible by 400.
3. Otherwise, check whether it is divisible by 4 but not by 100.
4. If either condition is true, it is a leap year.

```python
# Program to check leap year

year = int(input("Enter a year: "))

if year % 400 == 0:
    print("Leap year")
elif year % 4 == 0 and year % 100 != 0:
    print("Leap year")
else:
    print("Not a leap year")
```

Q7. Write a program to assign grades based on a student's marks.

Steps:
1. Enter student's marks.
2. Check the marks using if-elif-else.
3. Display the corresponding grade.

| Marks | Grade |
|---|---|
| 90–100 | A |
| 75–89 | B |
| 60–74 | C |
| 35–59 | D |
| 0–34 | Fail |

```python
# Program to assign grades based on marks

marks = int(input("Enter your marks: "))

if marks >= 90 and marks <= 100:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 35:
    print("Grade D")
elif marks >= 0:
    print("Fail")
else:
    print("Invalid marks")
```

Q9. Write a program to calculate an electricity bill using different usage slabs.

Electricity bills are calculated based on the number of units consumed.

For this example, we use sample rates:
| Units | Rate |
|---|---:|
| First 100 units | ₹1.50/unit |
| Next 100 units | ₹2.50/unit |
| Above 200 units | ₹4.00/unit |

```python
# Program to calculate electricity bill using slabs

units = int(input("Enter electricity units: "))

if units < 0:
    print("Invalid units")

elif units <= 100:
    bill = units * 1.50
    print("Electricity Bill: Rs.", bill)

elif units <= 200:
    bill = (100 * 1.50) + ((units - 100) * 2.50)
    print("Electricity Bill: Rs.", bill)

else:
    bill = (100 * 1.50) + (100 * 2.50) + ((units - 200) * 4.00)
    print("Electricity Bill: Rs.", bill)
```

Q10. What is a nested if statement? Give a suitable situation where it can be used.

A nested if statement means using one if statement inside another if statement.

It is used when we need to check multiple conditions.

Real-life example: Checking voting eligibility and voter registration.

```python
# Program to demonstrate nested if statement

age = int(input("Enter your age: "))

if age >= 18:
    registered = input("Are you registered to vote? (yes/no): ")

    if registered == "yes":
        print("You are eligible to vote")
    else:
        print("Please register to vote")
else:
    print("You are not eligible to vote")
```
